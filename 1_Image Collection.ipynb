```python
for label in labels:
    cap = cv2.VideoCapture(0)
    print('Collecting images for {}'.format(label))
    time.sleep(5)
    for imgnum in range(number_imgs):
        print('Collecting image {}'.format(imgnum))
        ret, frame = cap.read()
        imgname = os.path.join(IMAGES_PATH,label,label+'.'+'{}.jpg'.format(str(uuid.uuid1())))
        cv2.imwrite(imgname, frame)
        cv2.imshow('frame', frame)
        time.sleep(2)

        if cv2.waitKey(1) & 0xFF == ord('q'):
            break
cap.release()
cv2.destroyAllWindows()
```

### To check if the Folder name matches the Label name

```python
import os
import xml.etree.ElementTree as ET

ROOT_DIR = r"Tensorflow/workspace/images/collectedimages"

mismatches = []

for root, dirs, files in os.walk(ROOT_DIR):
    folder_name = os.path.basename(root)

    for file in files:
        if file.lower().endswith(".xml"):
            xml_path = os.path.join(root, file)

            try:
                tree = ET.parse(xml_path)
                xml_root = tree.getroot()

                objects = xml_root.findall("object")

                for obj in objects:
                    label = obj.find("name").text.strip()

                    if label != folder_name:
                        mismatches.append({
                            "xml": xml_path,
                            "folder": folder_name,
                            "label": label
                        })

            except Exception as e:
                print(f"Error reading {xml_path}: {e}")

if mismatches:
    print("\nMISMATCHES FOUND:\n")
    for item in mismatches:
        print(f"File   : {item['xml']}")
        print(f"Folder : {item['folder']}")
        print(f"Label  : {item['label']}")
        print("-" * 50)
else:
    print("All XML labels match their folder names.")
```

```python
import os
import shutil
import random

IMAGES_PATH = r"E:\Study\Projects\REAL-TIME OBJECT DETECTION FOR ASSISTING COMMUNICATION IN NON-VERBAL INDIVIDUALS\Tensorflow\workspace\images\collectedimages"
TRAIN_LABEL = r"E:\Study\Projects\REAL-TIME OBJECT DETECTION FOR ASSISTING COMMUNICATION IN NON-VERBAL INDIVIDUALS\Tensorflow\workspace\images\train_label"
TEST_LABEL  = r"E:\Study\Projects\REAL-TIME OBJECT DETECTION FOR ASSISTING COMMUNICATION IN NON-VERBAL INDIVIDUALS\Tensorflow\workspace\images\test_label"

os.makedirs(TRAIN_LABEL, exist_ok=True)
os.makedirs(TEST_LABEL,  exist_ok=True)

class_dir = os.path.join(IMAGES_PATH, 'A')

# Only labeled images
labeled = [f for f in os.listdir(class_dir)
           if f.endswith('.jpg') and
           os.path.exists(os.path.join(class_dir, f.replace('.jpg', '.xml')))]

random.shuffle(labeled)
split   = int(0.8 * len(labeled))
tr_imgs = labeled[:split]
te_imgs = labeled[split:]

for img in tr_imgs:
    shutil.copy(os.path.join(class_dir, img), TRAIN_LABEL)
    shutil.copy(os.path.join(class_dir, img.replace('.jpg', '.xml')), TRAIN_LABEL)

for img in te_imgs:
    shutil.copy(os.path.join(class_dir, img), TEST_LABEL)
    shutil.copy(os.path.join(class_dir, img.replace('.jpg', '.xml')), TEST_LABEL)

print(f'✅ Class A split done!')
print(f'   train_label/: {len(tr_imgs)} images')
print(f'   test_label/ : {len(te_imgs)} images')
```

# 6. Move them into a Training and Testing Partition (Manually)

```python
import os
import shutil
import random

root_dir = r"E:\Study\Projects\REAL-TIME OBJECT DETECTION FOR ASSISTING COMMUNICATION IN NON-VERBAL INDIVIDUALS\Tensorflow\workspace\images\collectedimages"
train_dir = os.path.join(root_dir, "train")
test_dir = os.path.join(root_dir, "test")

# Create train and test directories if they don't exist
os.makedirs(train_dir, exist_ok=True)
os.makedirs(test_dir, exist_ok=True)

for subfolder in os.listdir(root_dir):
    subfolder_path = os.path.join(root_dir, subfolder)
    if os.path.isdir(subfolder_path):
        images = [f for f in os.listdir(subfolder_path) if f.endswith(".jpg")]
        random.shuffle(images)
        
        split_index = int(0.8 * len(images))
        train_images = images[:split_index]
        test_images = images[split_index:]
        
        # Create subdirectories for train and test
        train_subdir = os.path.join(train_dir, subfolder)
        test_subdir = os.path.join(test_dir, subfolder)
        os.makedirs(train_subdir, exist_ok=True)
        os.makedirs(test_subdir, exist_ok=True)
        
        for img in train_images:
            img_path = os.path.join(subfolder_path, img)
            xml_path = os.path.join(subfolder_path, img.replace(".jpg", ".xml"))
            shutil.copy(img_path, train_subdir)
            if os.path.exists(xml_path):
                shutil.copy(xml_path, train_subdir)
        
        for img in test_images:
            img_path = os.path.join(subfolder_path, img)
            xml_path = os.path.join(subfolder_path, img.replace(".jpg", ".xml"))
            shutil.copy(img_path, test_subdir)
            if os.path.exists(xml_path):
                shutil.copy(xml_path, test_subdir)

print("Data split completed successfully!")
```
